# 🥗 NutriLens: 03 - Text Classification & Label Leakage Study
Evaluating category prediction and the empirical impact of label leakage.

```python
import sys
sys.path.append("..")
from python_src.preprocess import load_dataset
from python_src.classify import CategoryClassifier

foods = load_dataset("../data/food_nutrition_dataset.csv")
```

```python
clf_base = CategoryClassifier(leak_free=False)
res_base = clf_base.evaluate(foods)
print("Baseline Accuracy:", res_base["accuracy"], "Macro-F1:", res_base["macro_f1"])
```

```python
clf_leak = CategoryClassifier(leak_free=True)
res_leak = clf_leak.evaluate(foods)
print("Leak-Free Accuracy:", res_leak["accuracy"], "Macro-F1:", res_leak["macro_f1"])
```
