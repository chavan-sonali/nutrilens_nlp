# 🥗 NutriLens: 01 - Exploratory Data Analysis (EDA)
Dataset overview, schema validation, and nutrient distributions.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/food_nutrition_dataset.csv")
print("Dataset Shape:", df.shape)
print(df.info())
```

```python
print(df.describe())
print(df["category"].value_counts())
```

```python
plt.figure(figsize=(10, 5))
df["category"].value_counts().plot(kind="bar", color="#10B981")
plt.title("Food Items per Category")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x="calories_kcal", y="protein_g", hue="category")
plt.title("Protein vs Calories")
plt.show()
```
