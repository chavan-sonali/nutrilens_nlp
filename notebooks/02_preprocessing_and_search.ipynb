# 🥗 NutriLens: 02 - Preprocessing & Hybrid Retrieval
Tokenization, lemmatization, TF-IDF vectorization, and dense semantic search.

```python
import sys
sys.path.append("..")
from python_src.preprocess import load_dataset, clean_text, build_food_document
from python_src.search import TfidfSearchEngine, SemanticSearchEngine, hybrid_search
from python_src.nlu import calculate_quantiles, parse_query

foods = load_dataset("../data/food_nutrition_dataset.csv")
print(f"Loaded {len(foods)} foods")
```

```python
tfidf = TfidfSearchEngine().fit(foods)
sem = SemanticSearchEngine(foods)
print(f"TF-IDF Vocabulary Size: {len(tfidf.vocab)}")
```

```python
query = "high protein vegetarian food for weight loss"
combined, tfidf_s, sem_s = hybrid_search(tfidf, sem, query, alpha=0.4)
top_idx = sorted(range(len(foods)), key=lambda i: combined[i], reverse=True)[:5]
for r, idx in enumerate(top_idx, 1):
    print(f"#{r} {foods[idx]["food_name"]} ({foods[idx]["category"]}) - Score: {combined[idx]:.3f}")
```
